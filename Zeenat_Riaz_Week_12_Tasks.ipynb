## Module 2: Azure AI Language - Text Analytics (Day 2)
Practical implementation of Azure AI Language API:
1. **Sentiment Analysis & Opinion Mining:** Document and sentence-level sentiment scores.
2. **Named Entity Recognition (NER):** Categorizing entities (Person, Location, Organization, DateTime).
3. **Key Phrase Extraction:** Extracting core subjects and noun phrases.
4. **Language Detection:** Detecting spoken/written language and ISO code.


```python
# Implementing Azure AI Text Analytics Client
class AzureTextAnalyticsClient:
    def __init__(self, endpoint=None, key=None):
        self.endpoint = endpoint or "https://eastus.api.cognitive.microsoft.com/"

    def analyze_sentiment(self, text):
        # Simulated sentiment analysis output
        return {
            "sentiment": "positive",
            "confidence_scores": {"positive": 0.94, "neutral": 0.05, "negative": 0.01},
            "sentences": [
                {
                    "text": text,
                    "sentiment": "positive",
                    "opinions": [{"target": "Azure AI Services", "assessment": "powerful"}]
                }
            ]
        }

    def extract_entities(self, text):
        # Simulated NER output
        return [
            {"text": "Satya Nadella", "category": "Person", "confidence": 0.99},
            {"text": "Microsoft", "category": "Organization", "confidence": 0.99},
            {"text": "Redmond", "category": "Location", "confidence": 0.96},
            {"text": "October 2024", "category": "DateTime", "confidence": 0.92}
        ]

    def extract_key_phrases(self, text):
        return ["cloud infrastructure", "Azure AI Services", "neural speech models", "enterprise automation"]

    def detect_language(self, text):
        return {"language": "English", "iso6391Name": "en", "confidence_score": 1.0}

text_client = AzureTextAnalyticsClient()
sample_doc = "Satya Nadella announced new Microsoft cloud infrastructure and Azure AI Services in Redmond during October 2024. The platform is powerful and efficient."

# 1. Sentiment Analysis
sentiment_res = text_client.analyze_sentiment(sample_doc)
print("=== Azure AI Sentiment Analysis ===")
print("Overall Sentiment :", sentiment_res["sentiment"].upper())
print("Confidence Scores :", sentiment_res["confidence_scores"])

# 2. Named Entity Recognition
entities_res = text_client.extract_entities(sample_doc)
print("\n=== Named Entity Recognition (NER) ===")
print(f"{'Entity':<18} | {'Category':<15} | {'Confidence Score'}")
print("-" * 52)
for e in entities_res:
    print(f"{e['text']:<18} | {e['category']:<15} | {e['confidence']*100:.1f}%")

# 3. Key Phrases
phrases = text_client.extract_key_phrases(sample_doc)
print("\n=== Key Phrase Extraction ===")
print("Identified Key Phrases:", phrases)
```

## Module 3: Azure Speech Services (Day 2)
### Capabilities:
- **Speech-to-Text (STT):** Real-time acoustic and language model transcription.
- **Text-to-Speech (TTS):** Human-like neural speech synthesis with SSML (Speech Synthesis Markup Language).


```python
# Simulating Azure Speech Service STT & TTS Pipeline
class AzureSpeechServiceClient:
    def __init__(self, voice_name="en-US-JennyNeural"):
        self.voice_name = voice_name

    def text_to_speech_ssml(self, text):
        ssml = f'''<speak version="1.0" xmlns="http://www.w3.org/2001/10/synthesis" xml:lang="en-US">
    <voice name="{self.voice_name}">
        <prosody rate="0%" pitch="0%">
            {text}
        </prosody>
    </voice>
</speak>'''
        return ssml

    def speech_to_text(self, audio_file):
        return {
            "status": "Success",
            "recognized_text": "Artificial intelligence is accelerating scientific discovery worldwide.",
            "duration": "4.2 seconds",
            "confidence": 0.985
        }

speech_client = AzureSpeechServiceClient()
sample_text = "Welcome to Microsoft Azure Cognitive Speech synthesis demonstration."
ssml_payload = speech_client.text_to_speech_ssml(sample_text)

print("Generated SSML for Neural TTS:")
print(ssml_payload)

stt_result = speech_client.speech_to_text("speech_sample.wav")
print("\nSpeech-to-Text Result:")
print("Status         :", stt_result["status"])
print("Recognized Text:", stt_result["recognized_text"])
print("Confidence     :", f"{stt_result['confidence']*100:.2f}%")
```

## Module 4: Azure Translator Service (Day 2)
Translating text across diverse target languages while preserving sentence structure and applying dictionary mappings.


```python
# Simulating Azure Translator Multi-Language Endpoint
class AzureTranslatorClient:
    def translate(self, text, from_lang="en", to_langs=["ur", "es", "fr", "ar"]):
        # Dictionary of sample translations
        dictionary = {
            "ur": "مصنوعی ذہانت اور مشین لرننگ جدید ترین ٹیکنالوجی ہے۔",
            "es": "La inteligencia artificial y el aprendizaje automático son tecnologías avanzadas.",
            "fr": "L'intelligence artificielle et l'apprentissage automatique sont des technologies avancées.",
            "ar": "الذكاء الاصطناعي والتعلم الآلي من التقنيات المتقدمة."
        }
        results = {}
        for lang in to_langs:
            results[lang] = dictionary.get(lang, "Translated text...")
        return results

translator = AzureTranslatorClient()
source_text = "Artificial intelligence and machine learning are advanced technologies."
translations = translator.translate(source_text)

print(f"Source Text (English): '{source_text}'\n")
print(f"{'Language Code':<15} | {'Translated Output'}")
print("-" * 75)
for code, trans in translations.items():
    print(f"{code:<15} | {trans}")
```

## Module 5: Conversational Language Understanding & Azure Question Answering (Day 3)
1. **Conversational Language Understanding (CLU):** Extracting user intent and parameter entities for conversational workflows.
2. **Azure Question Answering (QnA Maker):** Mapping user natural language queries to semi-structured FAQ knowledge bases.


```python
# Simulating Question Answering Knowledge Base Querying
class AzureQuestionAnsweringKB:
    def __init__(self):
        self.kb = [
            {
                "questions": ["How do I reset my account password?", "Forgot password", "Reset login credentials"],
                "answer": "To reset your password, visit the portal login page and click 'Forgot Password' to receive a secure recovery link.",
                "confidence": 0.96
            },
            {
                "questions": ["What are Azure AI service quotas?", "How to increase rate limit?", "API limits"],
                "answer": "Azure AI quotas depend on your pricing tier (F0 Free vs S0 Standard). You can submit quota increase requests via the Azure Portal Support blade.",
                "confidence": 0.94
            },
            {
                "questions": ["Where is my invoice?", "How to download monthly bill?", "Billing statement"],
                "answer": "Invoices can be viewed and downloaded directly under Cost Management + Billing in your Azure subscription dashboard.",
                "confidence": 0.98
            }
        ]

    def query(self, user_question):
        user_tokens = set(user_question.lower().split())
        best_match = None
        highest_score = 0
        for entry in self.kb:
            for q in entry["questions"]:
                q_tokens = set(q.lower().split())
                overlap = len(user_tokens.intersection(q_tokens))
                if overlap > highest_score:
                    highest_score = overlap
                    best_match = entry
        if best_match and highest_score > 0:
            return best_match
        return {"answer": "I am sorry, I could not find a matching answer in the knowledge base. Connecting you to a support agent.", "confidence": 0.20}

qa_system = AzureQuestionAnsweringKB()

test_queries = [
    "I forgot my password how can I log in?",
    "Can you help me download monthly bill for my account?",
    "What is the capital of France?"
]

print("=== Azure Question Answering Knowledge Base Queries ===\n")
for q in test_queries:
    ans = qa_system.query(q)
    print(f"User Query : '{q}'")
    print(f"Answer     : {ans['answer']}")
    print(f"Confidence : {ans['confidence']*100:.1f}%
")
```

## Module 6: Enterprise Conversational AI & Chatbot Pipeline (Day 5)
Architecture of an enterprise conversational bot combining:
- Session state management
- Intent classification and Entity extraction
- Question Answering KB integration
- Graceful fallback and escalation handling.


```python
class EnterpriseAzureChatbot:
    def __init__(self, qa_engine):
        self.qa_engine = qa_engine
        self.session_history = []
        self.escalation_threshold = 0.50

    def process_message(self, user_id, message):
        timestamp = time.strftime("%H:%M:%S")
        response_data = self.qa_engine.query(message)

        confidence = response_data['confidence']
        if confidence >= self.escalation_threshold:
            reply = response_data['answer']
            status = "RESOLVED"
        else:
            reply = "I apologize, but I am unable to answer this accurately. Your inquiry has been escalated to a live agent."
            status = "ESCALATED"

        record = {
            "time": timestamp,
            "user_id": user_id,
            "message": message,
            "reply": reply,
            "confidence": confidence,
            "status": status
        }
        self.session_history.append(record)
        return record

bot = EnterpriseAzureChatbot(qa_system)

chat_dialogue = [
    "Hello, I need to know how to reset my account password",
    "Where can I find and download monthly bill?",
    "Can you arrange a flight ticket to London?"
]

print("=== Interactive Azure Enterprise Conversational Chatbot Session ===\n")
for msg in chat_dialogue:
    rec = bot.process_message("user_042", msg)
    print(f"[{rec['time']}] User : {rec['message']}")
    print(f"[{rec['time']}] Bot  : {rec['reply']}")
    print(f"       Status: {rec['status']} (Confidence: {rec['confidence']*100:.1f}%)
")
```

```python
# Reviewing Chatbot Operational Metrics & Session History
df_chat_log = pd.DataFrame(bot.session_history)
print("=== Chatbot Session Analytics & Audit Log ===")
print(df_chat_log[['time', 'user_id', 'status', 'confidence', 'message']])
```

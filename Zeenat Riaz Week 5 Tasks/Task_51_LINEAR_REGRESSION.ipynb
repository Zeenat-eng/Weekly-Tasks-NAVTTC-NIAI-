```python
"""
Task 51: Machine Learning - Implementation of Linear Regression
Week: 5
Reference: https://stackabuse.com/linear-regression-in-python-with-scikit-learn/
"""

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics

print("=" * 65)
print("     MACHINE LEARNING: SIMPLE & MULTIPLE LINEAR REGRESSION")
print("=" * 65)

# PART 1: SIMPLE LINEAR REGRESSION (1 Feature -> 1 Target)
# Scenario: Predicting Student Exam Score based on Study Hours

print("\n--- PART 1: SIMPLE LINEAR REGRESSION ---")

# 1. Dataset Preparation
data_simple = {
    'Hours': [2.5, 5.1, 3.2, 8.5, 3.5, 1.5, 9.2, 5.5, 8.3, 2.7, 7.7, 5.9, 4.5, 3.3, 1.1, 8.9, 2.5, 1.9, 6.1, 7.4],
    'Scores': [21, 47, 27, 75, 30, 20, 88, 60, 81, 25, 85, 62, 41, 42, 17, 95, 30, 24, 67, 69]
}
df_simple = pd.DataFrame(data_simple)
print("Dataset Head:")
print(df_simple.head())

# Features (X) and Target (y)
X_simple = df_simple[['Hours']]  # 2D DataFrame for Scikit-Learn
y_simple = df_simple['Scores']

# 2. Splitting into Train and Test sets (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(X_simple, y_simple, test_size=0.2, random_state=42)

# 3. Model Training
simple_model = LinearRegression()
simple_model.fit(X_train, y_train)

# 4. Model Coefficients
slope = simple_model.coef_[0]
intercept = simple_model.intercept_
print(f"\nFitted Linear Equation: Score = ({slope:.2f} * Hours) + ({intercept:.2f})")

# 5. Making Predictions
y_pred = simple_model.predict(X_test)

comparison = pd.DataFrame({'Actual': y_test, 'Predicted': y_pred.round(2)})
print("\nPrediction on Test Set:")
print(comparison)

# 6. Model Evaluation Metrics
mae = metrics.mean_absolute_error(y_test, y_pred)
mse = metrics.mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = metrics.r2_score(y_test, y_pred)

print("\nSimple Linear Regression Evaluation:")
print(f"  - Mean Absolute Error (MAE)  : {mae:.2f}")
print(f"  - Mean Squared Error (MSE)   : {mse:.2f}")
print(f"  - Root Mean Squared Error    : {rmse:.2f}")
print(f"  - R-squared Score (R2)       : {r2:.4f} ({r2*100:.1f}% variance explained)")

# PART 2: MULTIPLE LINEAR REGRESSION (Multiple Features -> 1 Target)
# Scenario: House Price Prediction (Size, Bedrooms, Age -> Price in $1000s)

print("\n" + "=" * 65)
print("--- PART 2: MULTIPLE LINEAR REGRESSION ---")

data_multi = {
    'Size_sqft': [1500, 1800, 2400, 3000, 3500, 1200, 2000, 2800, 1600, 2200],
    'Bedrooms':  [3, 3, 4, 4, 5, 2, 3, 4, 3, 3],
    'Age_years': [10, 15, 5, 2, 1, 20, 8, 4, 12, 6],
    'Price_k':   [300, 340, 450, 560, 650, 240, 380, 520, 310, 410]
}
df_multi = pd.DataFrame(data_multi)

X_m = df_multi[['Size_sqft', 'Bedrooms', 'Age_years']]
y_m = df_multi['Price_k']

X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(X_m, y_m, test_size=0.2, random_state=42)

multi_model = LinearRegression()
multi_model.fit(X_train_m, y_train_m)

print("Feature Coefficients:")
for col, coef in zip(X_m.columns, multi_model.coef_):
    print(f"  - {col:<12}: {coef:+.4f}")
print(f"Intercept     : {multi_model.intercept_:.2f}")

y_pred_m = multi_model.predict(X_test_m)
print(f"\nMultiple Regression R2 Score: {metrics.r2_score(y_test_m, y_pred_m):.4f}")
print("=" * 65)

```

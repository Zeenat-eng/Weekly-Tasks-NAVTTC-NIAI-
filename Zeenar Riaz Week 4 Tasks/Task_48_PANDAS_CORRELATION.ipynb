```python
"""
Task 48: Exercise on Pandas Data Correlation
Week: 4
Reference: https://www.w3schools.com/python/pandas_correlations.asp
"""

import pandas as pd

print("=" * 60)
print("             PANDAS: DATA CORRELATION")
print("=" * 60)

# Sample health and workout metrics dataset
fitness_data = {
    'Duration': [60, 60, 60, 45, 45, 60, 60, 45, 30, 60],
    'Pulse': [110, 117, 103, 109, 117, 102, 110, 104, 109, 98],
    'Maxpulse': [130, 145, 135, 175, 148, 127, 136, 134, 133, 124],
    'Calories': [409, 479, 340, 282, 406, 300, 374, 253, 195, 269]
}

df = pd.DataFrame(fitness_data)
print("Fitness Dataset:")
print(df)

# Calculate Correlation Matrix using .corr() method (Pearson correlation)
corr_matrix = df.corr()
print("\nCorrelation Matrix (df.corr()):")
print(corr_matrix)

print("\n" + "=" * 60)
print("CORRELATION INTERPRETATION GUIDE:")
print("1. Value ranges between -1.0 and +1.0.")
print("2. 1.0 indicates a PERFECT positive correlation (e.g. Duration vs Duration).")
print("3. Higher positive value (e.g. 0.9) means when one increases, the other increases.")
print("4. Higher negative value (e.g. -0.9) means when one increases, the other decreases.")
print("5. Value close to 0.0 (e.g. 0.05 or -0.09) means NO linear relationship.")
print("=" * 60)

```

###  TASK 47 PANDAS CLEANING DATA


```python
"""
Task 47: Exercise on Pandas Data Cleaning Techniques
Week: 4
Reference: https://www.w3schools.com/python/pandas_cleaning.asp
"""

import pandas as pd
import numpy as np

print("=" * 60)
print("             PANDAS: DATA CLEANING TECHNIQUES")
print("=" * 60)

# Dirty Dataset with:
# - Missing values (NaN / None)
# - Duplicate rows
# - Wrong format / incorrect values
dirty_data = {
    'ID': [1, 2, 3, 4, 5, 5],
    'Duration': [60, 60, np.nan, 45, 450, 45],  # 450 is a wrong outlier
    'Date': ['2026/10/01', '2026/10/02', '20261003', '2026/10/04', '2026/10/05', '2026/10/04'],
    'Calories': [409.1, 479.0, 340.0, np.nan, 406.0, np.nan]
}

df = pd.DataFrame(dirty_data)
print("Raw / Dirty DataFrame:")
print(df)

# --- 1. Handling Missing Data ---
# A: Check for null values
print(f"\nMissing values per column:\n{df.isnull().sum()}")

# B: Fill missing Calories with Mean value
mean_calories = df['Calories'].mean()
df['Calories'] = df['Calories'].fillna(mean_calories)
print(f"\nAfter filling missing Calories with mean ({mean_calories:.2f}):")
print(df)

# C: Drop rows with remaining NaN values
df.dropna(subset=['Duration'], inplace=True)

# --- 2. Correcting Wrong Data Formats ---
df['Date'] = pd.to_datetime(df['Date'], format='mixed')
print("\nAfter converting 'Date' to datetime object:")
print(df)

# --- 3. Fixing Outliers / Wrong Values ---
# If duration is greater than 120, cap it at 60
df.loc[df['Duration'] > 120, 'Duration'] = 60
print("\nAfter correcting outlier Duration > 120:")
print(df)

# --- 4. Removing Duplicates ---
print(f"\nDuplicate rows found: {df.duplicated().sum()}")
df.drop_duplicates(inplace=True)
print("\nFinal Cleaned DataFrame:")
print(df)

```

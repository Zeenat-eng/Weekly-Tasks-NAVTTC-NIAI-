###  TASK 44 PANDAS DATAFRAMES


```python
"""
Task 44: Exercise on Pandas DataFrame
Week: 4
Reference: https://www.w3schools.com/python/pandas_dataframes.asp
"""

import pandas as pd

print("=" * 60)
print("             PANDAS: DATAFRAMES & INDEXING")
print("=" * 60)

# 1. Creating a DataFrame from Dictionary of Lists
data = {
    "Calories": [420, 380, 390],
    "Duration": [50, 40, 45]
}
df = pd.DataFrame(data)
print("DataFrame with default numeric index:")
print(df)

# 2. Locate Row using loc attribute (label-based)
print("\nLocate single row via df.loc[0]:")
print(df.loc[0])

print("\nLocate multiple rows via df.loc[[0, 1]]:")
print(df.loc[[0, 1]])

# 3. Named Indexes (Custom Row Labels)
df_custom = pd.DataFrame(data, index=["day1", "day2", "day3"])
print("\nDataFrame with Named Indexes:")
print(df_custom)

print("\nLocate named row via df.loc['day2']:")
print(df_custom.loc["day2"])

# 4. Integer-location based indexing using iloc
print("\nLocate row via df.iloc[1] (Position-based):")
print(df_custom.iloc[1])

```

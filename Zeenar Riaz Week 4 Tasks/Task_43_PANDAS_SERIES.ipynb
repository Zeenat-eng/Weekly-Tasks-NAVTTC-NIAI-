###  TASK 43 PANDAS SERIES


```python
"""
Task 43: Exercise on Pandas Series Data
Week: 4
Reference: https://www.w3schools.com/python/pandas_series.asp
"""

import pandas as pd

print("=" * 60)
print("             PANDAS: SERIES DATA STRUCTURE")
print("=" * 60)

# 1. Simple Series from Python List
a = [1, 7, 2, 9]
my_series = pd.Series(a)
print("1. Default Series (with default 0-based integer index):")
print(my_series)
print(f"First element: {my_series[0]}")

# 2. Series with Custom Labels (Index)
custom_series = pd.Series(a, index=["x", "y", "z", "w"])
print("\n2. Series with Custom Labels:")
print(custom_series)
print(f"Access element with label 'y': {custom_series['y']}")

# 3. Series from Key/Value Object (Dictionary)
calories = {"day1": 420, "day2": 380, "day3": 390, "day4": 410}
cal_series = pd.Series(calories)
print("\n3. Series created from a Dictionary:")
print(cal_series)

# 4. Selecting Specific Keys using index parameter
selected_cals = pd.Series(calories, index=["day1", "day3"])
print("\n4. Filtered Series using specific index keys:")
print(selected_cals)

```

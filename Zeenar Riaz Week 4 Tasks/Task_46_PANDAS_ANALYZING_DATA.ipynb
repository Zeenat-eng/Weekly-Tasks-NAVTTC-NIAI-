###  TASK 46 PANDAS ANALYZING DATA


```python
"""
Task 46: Exercise on Pandas Data Analyzing
Week: 4
Reference: https://www.w3schools.com/python/pandas_analyzing.asp
"""

import pandas as pd

print("=" * 60)
print("             PANDAS: ANALYZING DATA")
print("=" * 60)

# Sample dataset
data = {
    'Age': [25, 30, 45, 35, 22, 48, 52, 40, 28, 33],
    'Salary': [50000, 62000, 85000, 71000, 42000, 93000, 105000, 80000, 58000, 67000],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'IT', 'Finance', 'Executive', 'HR', 'IT', 'Finance'],
    'Experience_Years': [2, 5, 15, 8, 1, 18, 22, 12, 4, 7]
}
df = pd.DataFrame(data)

# 1. Viewing first N rows using head()
print("1. First 5 Rows (df.head()):")
print(df.head())

# 2. Viewing last N rows using tail()
print("\n2. Last 3 Rows (df.tail(3)):")
print(df.tail(3))

# 3. General Information about Dataset using info()
print("\n3. Dataset Information (df.info()):")
df.info()

# 4. Statistical Summary using describe()
print("\n4. Summary Statistics for Numerical Columns (df.describe()):")
print(df.describe())

# 5. Value counts for Categorical Columns
print("\n5. Department Frequency Count (value_counts):")
print(df['Department'].value_counts())

```

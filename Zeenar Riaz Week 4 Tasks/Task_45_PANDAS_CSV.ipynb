```python
"""
Task 45: Exercise on Pandas Reading & Writing CSV Files
Week: 4
Reference: https://www.w3schools.com/python/pandas_csv.asp
"""

import os
import pandas as pd

print("=" * 60)
print("             PANDAS: READING & WRITING CSV")
print("=" * 60)

csv_filename = "sample_health_data.csv"

# 1. Create a sample CSV file if it doesn't already exist
sample_content = """Duration,Pulse,Maxpulse,Calories
60,110,130,409.1
60,117,145,479.0
60,103,135,340.0
45,109,175,282.4
45,117,148,406.0
60,102,127,300.0
60,110,136,374.0
45,104,134,253.3
30,109,133,195.1
60,98,124,269.0
"""

with open(csv_filename, "w", encoding="utf-8") as f:
    f.write(sample_content.strip())
print(f"Sample CSV file '{csv_filename}' verified/created.")

# 2. Read CSV file into a DataFrame
df = pd.read_csv(csv_filename)

print("\nDisplaying DataFrame using df.to_string():")
print(df.to_string())

# 3. Checking Pandas Display Options
print(f"\nDefault max_rows displayed by Pandas: {pd.options.display.max_rows}")

# 4. Exporting to a new CSV file
output_csv = "exported_output.csv"
df.to_csv(output_csv, index=False)
print(f"DataFrame successfully exported to '{output_csv}'.")

# Clean up exported copy
if os.path.exists(output_csv):
    os.remove(output_csv)

```

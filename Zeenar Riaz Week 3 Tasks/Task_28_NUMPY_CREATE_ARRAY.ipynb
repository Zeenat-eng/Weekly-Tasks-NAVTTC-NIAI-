```python
"""
Task 28: Exercise on NumPy Creating Arrays
Week: 3
Reference: https://www.w3schools.com/python/numpy_creating_arrays.asp
"""

import numpy as np

print("=" * 60)
print("             NUMPY: CREATING ARRAYS")
print("=" * 60)

# 1. 0-D Array (Scalar)
arr_0d = np.array(42)
print(f"0-D Array: {arr_0d}, Dimensions: {arr_0d.ndim}")

# 2. 1-D Array (Vector)
arr_1d = np.array([1, 2, 3, 4, 5])
print(f"\n1-D Array: {arr_1d}, Dimensions: {arr_1d.ndim}")

# 3. 2-D Array (Matrix)
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print(f"\n2-D Array:\n{arr_2d}\nDimensions: {arr_2d.ndim}")

# 4. 3-D Array (Tensor)
arr_3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(f"\n3-D Array:\n{arr_3d}\nDimensions: {arr_3d.ndim}")

# 5. Higher Dimensional Arrays (ndmin parameter)
arr_5d = np.array([1, 2, 3, 4], ndmin=5)
print(f"\nArray with ndmin=5: {arr_5d}")
print(f"Number of dimensions: {arr_5d.ndim}")

# 6. Built-in Creation Routines
zeros = np.zeros((2, 3))
ones = np.ones((2, 3))
range_arr = np.arange(0, 10, 2)
linear_spaced = np.linspace(0, 1, 5)

print("\nBuilt-in Creation Routines:")
print(f"np.zeros((2, 3)):\n{zeros}")
print(f"np.ones((2, 3)):\n{ones}")
print(f"np.arange(0, 10, 2): {range_arr}")
print(f"np.linspace(0, 1, 5): {linear_spaced}")

```

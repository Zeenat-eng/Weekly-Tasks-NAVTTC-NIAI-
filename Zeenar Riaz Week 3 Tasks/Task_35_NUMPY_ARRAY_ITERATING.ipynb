```python
"""
Task 35: Exercise on NumPy Array Iteration
Week: 3
Reference: https://www.w3schools.com/python/numpy_array_iterating.asp
"""

import numpy as np

print("=" * 60)
print("             NUMPY: ARRAY ITERATION")
print("=" * 60)

# 1. Iterating 1-D Array
arr_1d = np.array([1, 2, 3])
print("Iterating 1-D Array:")
for x in arr_1d:
    print(f"  Item: {x}")

# 2. Iterating 2-D Array (Rows vs Scalars)
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print("\nIterating 2-D Array by Row:")
for row in arr_2d:
    print(f"  Row: {row}")

print("\nIterating each scalar in 2-D Array with nested loops:")
for row in arr_2d:
    for item in row:
        print(item, end=" ")
print()

# 3. Iterating using np.nditer() (Advanced & Fast)
print("\nIterating 3-D Array elements using np.nditer():")
arr_3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
for val in np.nditer(arr_3d):
    print(val, end=" ")
print()

# 4. Iterating with ndenumerate() to get index and value
print("\nEnumerating array elements with coordinates (ndenumerate):")
arr_enum = np.array([[10, 20], [30, 40]])
for idx, val in np.ndenumerate(arr_enum):
    print(f"  Index {idx} => Value {val}")

```

###  TASK 38 NUMPY ARRAY SEARCH


```python
"""
Task 38: Exercise on NumPy Array Searching
Week: 3
Reference: https://www.w3schools.com/python/numpy_array_search.asp
"""

import numpy as np

print("=" * 60)
print("             NUMPY: SEARCHING ARRAYS")
print("=" * 60)

arr = np.array([1, 2, 3, 4, 5, 4, 4, 8, 9])
print(f"Array: {arr}")

# 1. np.where() - returns tuple of indices where condition is True
where_4 = np.where(arr == 4)
print(f"\nIndices where value is 4: {where_4[0]}")

# Finding indices of all even numbers
even_indices = np.where(arr % 2 == 0)
print(f"Indices of even numbers: {even_indices[0]}")
print(f"Even values: {arr[even_indices]}")

# Finding indices of all odd numbers
odd_indices = np.where(arr % 2 == 1)
print(f"Indices of odd numbers: {odd_indices[0]}")

# 2. np.searchsorted() - binary search for insertion in sorted array
sorted_arr = np.array([6, 7, 8, 9, 10])
print(f"\nSorted Array: {sorted_arr}")
insert_pos = np.searchsorted(sorted_arr, 8)
print(f"Position where 8 should be inserted: {insert_pos}")

# Searching multiple values for insertion
multi_pos = np.searchsorted(sorted_arr, [5, 7, 11])
print(f"Positions for [5, 7, 11]: {multi_pos}")

```

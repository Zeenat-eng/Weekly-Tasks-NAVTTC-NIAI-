###  TASK 36 NUMPY ARRAY JOIN


```python
"""
Task 36: Exercise on NumPy Matrix Joining
Week: 3
Reference: https://www.w3schools.com/python/numpy_array_join.asp
"""

import numpy as np

print("=" * 60)
print("             NUMPY: JOINING ARRAYS")
print("=" * 60)

arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])

print(f"Array 1:\n{arr1}")
print(f"Array 2:\n{arr2}")

# 1. Joining along axis 0 (Rows / vertically)
concat_axis0 = np.concatenate((arr1, arr2), axis=0)
print(f"\nConcatenate along axis=0:\n{concat_axis0}")

# 2. Joining along axis 1 (Columns / horizontally)
concat_axis1 = np.concatenate((arr1, arr2), axis=1)
print(f"\nConcatenate along axis=1:\n{concat_axis1}")

# 3. Stacking Functions
# hstack() stacks horizontally (along columns)
print(f"\nHorizontal Stack (hstack):\n{np.hstack((arr1, arr2))}")

# vstack() stacks vertically (along rows)
print(f"\nVertical Stack (vstack):\n{np.vstack((arr1, arr2))}")

# dstack() stacks along height / depth
print(f"\nDepth Stack (dstack) shape: {np.dstack((arr1, arr2)).shape}")

```

```python
"""
Task 31: Exercise on NumPy Data Types & Type Conversion
Week: 3
Reference: https://www.w3schools.com/python/numpy_data_types.asp
"""

import numpy as np

print("=" * 60)
print("             NUMPY: DATA TYPES & ASTYPE")
print("=" * 60)

# 1. Checking Data Types of Arrays
arr_int = np.array([1, 2, 3, 4])
arr_float = np.array([1.5, 2.5, 3.5])
arr_str = np.array(['apple', 'banana', 'cherry'])

print(f"Integer array dtype : {arr_int.dtype}")
print(f"Float array dtype   : {arr_float.dtype}")
print(f"String array dtype  : {arr_str.dtype}")

# 2. Creating Arrays with Defined Data Type
arr_defined = np.array([1, 2, 3, 4], dtype='S')  # string bytes
print(f"\nArray created with dtype='S': {arr_defined}")
print(f"Data type: {arr_defined.dtype}")

arr_int4 = np.array([1, 2, 3, 4], dtype='i4')   # 4-byte integer (int32)
print(f"Array created with dtype='i4': {arr_int4.dtype}")

# 3. Converting Data Type Using astype()
original_floats = np.array([1.1, 2.7, 3.9, 4.2])
converted_ints = original_floats.astype('i')     # converts to integer
print(f"\nOriginal float array : {original_floats}")
print(f"Converted to int (i) : {converted_ints}")
print(f"New data type        : {converted_ints.dtype}")

# Converting Integer to Boolean
arr_bool_source = np.array([1, 0, 3, 0, -5])
arr_bool = arr_bool_source.astype(bool)
print(f"\nOriginal numbers: {arr_bool_source}")
print(f"Converted to bool: {arr_bool}")

```

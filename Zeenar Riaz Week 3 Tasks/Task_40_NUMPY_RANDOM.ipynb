```python
"""
Task 40: Exercise on NumPy Random Techniques
Week: 3
Reference: https://www.w3schools.com/python/numpy_random.asp
"""

import numpy as np
from numpy import random

print("=" * 60)
print("             NUMPY: RANDOM MODULE")
print("=" * 60)

# Set seed for reproducible results
random.seed(42)

# 1. Generate Random Integers
rand_int = random.randint(100)
print(f"Single random int between 0 and 100: {rand_int}")

rand_int_array = random.randint(10, 50, size=(5,))
print(f"1-D array of random ints (10 to 50): {rand_int_array}")

rand_2d = random.randint(1, 20, size=(3, 3))
print(f"2-D array (3x3) of random ints:\n{rand_2d}")

# 2. Generate Random Floats between 0 and 1
rand_float = random.rand()
print(f"\nSingle random float: {rand_float:.4f}")

rand_float_arr = random.rand(4)
print(f"Array of 4 random floats: {rand_float_arr}")

# 3. Random Choice from Array
choices = [10, 20, 30, 40, 50]
chosen = random.choice(choices)
print(f"\nRandom choice from {choices}: {chosen}")

# Generate array based on probability distribution (p)
prob_arr = random.choice([3, 5, 7, 9], p=[0.1, 0.3, 0.6, 0.0], size=(10,))
print(f"Array with custom probabilities: {prob_arr}")

# 4. Shuffle vs Permutation
arr_to_shuffle = np.array([1, 2, 3, 4, 5])
random.shuffle(arr_to_shuffle)  # Modifies original in-place
print(f"\nShuffled in-place: {arr_to_shuffle}")

arr_perm = np.array([10, 20, 30, 40, 50])
new_perm = random.permutation(arr_perm)  # Leaves original unchanged
print(f"Permuted copy   : {new_perm}")
print(f"Original kept   : {arr_perm}")

```

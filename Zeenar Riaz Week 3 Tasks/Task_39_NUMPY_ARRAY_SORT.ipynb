###  TASK 39 NUMPY ARRAY SORT


```python
"""
Task 39: Exercise on NumPy Array Sorting
Week: 3
Reference: https://www.w3schools.com/python/numpy_array_sort.asp
"""

import numpy as np

print("=" * 60)
print("             NUMPY: SORTING ARRAYS")
print("=" * 60)

# 1. Sorting 1-D Numeric Array
num_arr = np.array([3, 2, 0, 1, 5, -4, 8])
sorted_nums = np.sort(num_arr)
print(f"Original numbers : {num_arr}")
print(f"Sorted numbers   : {sorted_nums}")

# 2. Sorting Alphabetical Array
str_arr = np.array(['banana', 'cherry', 'apple', 'date'])
print(f"\nOriginal strings : {str_arr}")
print(f"Sorted strings   : {np.sort(str_arr)}")

# 3. Sorting Boolean Array
bool_arr = np.array([True, False, True, False, True])
print(f"\nOriginal booleans: {bool_arr}")
print(f"Sorted booleans  : {np.sort(bool_arr)} (False comes before True)")

# 4. Sorting 2-D Array
arr_2d = np.array([[3, 2, 4], [5, 0, 1]])
print(f"\nOriginal 2-D Array:\n{arr_2d}")
# Default sorts along the last axis (axis 1 - each row sorted internally)
print(f"Sorted 2-D Array along axis=1:\n{np.sort(arr_2d, axis=1)}")
# Sorting along axis 0 (columns)
print(f"Sorted 2-D Array along axis=0:\n{np.sort(arr_2d, axis=0)}")

```

###  TASK 32 NUMPY COPY VS VIEW


```python
"""
Task 32: Exercise on NumPy Copy vs View
Week: 3
Reference: https://www.w3schools.com/python/numpy_copy_vs_view.asp
"""

import numpy as np

print("=" * 60)
print("             NUMPY: COPY vs VIEW")
print("=" * 60)

# 1. Demonstrating COPY (Owns data; changes don't affect original)
arr = np.array([1, 2, 3, 4, 5])
arr_copy = arr.copy()
arr_copy[0] = 99

print("--- COPY DEMONSTRATION ---")
print(f"Original array : {arr}")
print(f"Copied array   : {arr_copy}")
print("Change in copy did NOT affect the original array.")

# 2. Demonstrating VIEW (Does NOT own data; changes affect original)
arr_view = arr.view()
arr_view[0] = 77

print("\n--- VIEW DEMONSTRATION ---")
print(f"Original array : {arr}")
print(f"View array     : {arr_view}")
print("Change in view directly altered the original array!")

# 3. Checking Ownership Using .base Attribute
# If an array owns the data, base is None. Otherwise, base refers to original object.
print("\n--- CHECKING OWNERSHIP VIA .base ---")
print(f"arr_copy.base : {arr_copy.base} (None means it OWNS the data)")
print(f"arr_view.base : {arr_view.base} (Points to original array)")

```

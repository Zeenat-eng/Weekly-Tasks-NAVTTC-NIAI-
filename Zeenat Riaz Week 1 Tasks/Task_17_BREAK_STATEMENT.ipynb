###  TASK 17 BREAK STATEMENT


```python
"""
Task 17: Exercise on Break Statement in Python
Week: 1
Reference: https://www.geeksforgeeks.org/python-break-statement/
"""

print("--- 1. Break in a For Loop ---")
for num in range(1, 10):
    if num == 5:
        print(f"Reached {num}: breaking out of loop.")
        break
    print(f"Number: {num}")
print("Outside loop.")

print("\n--- 2. Break in a While Loop ---")
val = 10
while val > 0:
    print(f"val = {val}")
    val -= 2
    if val == 4:
        print("Condition val == 4 met, breaking.")
        break

print("\n--- 3. Searching for an Element with Break ---")
data = [12, 45, 78, 23, 89, 99, 105]
target = 23
found = False

for index, val in enumerate(data):
    if val == target:
        print(f"Target {target} found at index {index}!")
        found = True
        break
if not found:
    print("Target not found.")

print("\n--- 4. Breaking from Nested Loops ---")
matrix = [
    [1, 2, 3],
    [4, 999, 6],
    [7, 8, 9]
]
target_found = False
for r_idx, row in enumerate(matrix):
    for c_idx, cell in enumerate(row):
        if cell == 999:
            print(f"Found magic number 999 at row {r_idx}, col {c_idx}!")
            target_found = True
            break
    if target_found:
        break

```

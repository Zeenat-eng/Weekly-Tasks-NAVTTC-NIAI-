###  TASK 16 WHILE LOOPS


```python
"""
Task 16: Exercise on While Loops in Python
Week: 1
Reference: https://www.geeksforgeeks.org/python-while-loops/
"""

print("--- 1. Basic While Loop Counter ---")
count = 1
while count <= 5:
    print(f"Count is: {count}")
    count += 1

print("\n--- 2. Sum of First N Natural Numbers ---")
n = 10
total_sum = 0
current = 1
while current <= n:
    total_sum += current
    current += 1
print(f"Sum of numbers from 1 to {n} is: {total_sum}")

print("\n--- 3. While Loop with else Block ---")
counter = 0
while counter < 3:
    print(f"Inside while loop: {counter}")
    counter += 1
else:
    print("Inside else block (loop condition became false)")

print("\n--- 4. Sentinel-Controlled Loop Simulation ---")
items = ["apple", "banana", "STOP", "orange"]
idx = 0
while idx < len(items):
    item = items[idx]
    if item == "STOP":
        print(f"Encountered sentinel '{item}', stopping processing.")
        break
    print(f"Processing item: {item}")
    idx += 1

```

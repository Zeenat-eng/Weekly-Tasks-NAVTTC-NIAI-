```python
"""
Task 1: Software Installation Guide & Environment Verification
Course: Artificial Intelligence, Machine Learning & Deep Learning
Week: 1

This task covers the installation and environment setup for:
1. Anaconda3 (Python Distribution)
2. PyTorch (Deep Learning Framework)
3. TensorFlow 2.0 (Deep Learning Framework)
4. VSCode (Visual Studio Code Editor)
5. PyCharm (Python IDE)
"""

import sys
import platform

def verify_environment():
    print("=" * 60)
    print("      ENVIRONMENT & SOFTWARE VERIFICATION TOOL")
    print("=" * 60)
    print(f"Operating System : {platform.system()} {platform.release()} ({platform.architecture()[0]})")
    print(f"Python Executable: {sys.executable}")
    print(f"Python Version   : {sys.version.split()[0]}")
    print("-" * 60)

    packages = {
        "torch": "PyTorch",
        "tensorflow": "TensorFlow 2.0+",
        "numpy": "NumPy",
        "pandas": "Pandas",
        "matplotlib": "Matplotlib",
        "sklearn": "Scikit-Learn",
        "scipy": "SciPy"
    }

    print("Checking Installed Packages:")
    for pkg, label in packages.items():
        try:
            mod = __import__(pkg)
            version = getattr(mod, "__version__", "Installed")
            print(f"  [OK] {label:<20} : Version {version}")
        except ImportError:
            print(f"  [--] {label:<20} : Not Installed")

    print("-" * 60)
    print("\nINSTALLATION INSTRUCTIONS:")
    print("1. Anaconda3: Download from https://www.anaconda.com/download")
    print("2. VSCode:    Download from https://code.visualstudio.com/")
    print("3. PyCharm:   Download from https://www.jetbrains.com/pycharm/")
    print("4. PyTorch:   Run in terminal: pip install torch torchvision torchaudio")
    print("5. TensorFlow: Run in terminal: pip install tensorflow")
    print("=" * 60)

if __name__ == "__main__":
    verify_environment()

```

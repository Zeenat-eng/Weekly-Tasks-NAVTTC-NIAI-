###  TASK 20 USER DEFINED FUNCTIONS


```python
"""
Task 20: Exercise on User-Defined Functions in Python
Weeks: 1
Reference: https://www.geeksforgeeks.org/functions-in-python/
"""

# 1. Simple Function with Return Value
def greet_user(name):
    """Function to display a welcome greeting."""
    return f"Hello, {name}! Welcome to the AI & Machine Learning Course."

# 2. Function with Default Arguments
def power(base, exponent=2):
    """Calculates base raised to exponent (default square)."""
    return base ** exponent

# 3. Function with Arbitrary Positional Arguments (*args)
def calculate_sum(*numbers):
    """Calculates sum of variable number of arguments."""
    total = 0
    for num in numbers:
        total += num
    return total

# 4. Function with Keyword Arguments (**kwargs)
def student_profile(**details):
    """Displays key-value details of a student."""
    print("Student Profile:")
    for key, value in details.items():
        print(f"  {key.capitalize()}: {value}")

# 5. Recursive Function (Factorial)
def factorial(n):
    """Calculates factorial recursively."""
    if n <= 1:
        return 1
    return n * factorial(n - 1)

# 6. Lambda Functions (Anonymous functions)
cube = lambda x: x ** 3

if __name__ == "__main__":
    print(greet_user("Austin"))
    print(f"5^2 = {power(5)}")
    print(f"2^4 = {power(2, 4)}")
    print(f"Sum of (10, 20, 30, 40) = {calculate_sum(10, 20, 30, 40)}")
    student_profile(name="Zeenat Riaz", course="AI & ML", institute="NAVTTC/NIAI", year=2026)
    print(f"Factorial of 5 = {factorial(5)}")
    print(f"Cube of 4 = {cube(4)}")

```

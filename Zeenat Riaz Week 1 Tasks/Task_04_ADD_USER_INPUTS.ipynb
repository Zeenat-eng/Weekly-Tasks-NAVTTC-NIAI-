###  TASK 04 ADD USER INPUTS


```python
# Task 4: Python Program to Store Input Numbers and Add
# Week: 1

# Store input numbers
# If running non-interactively or with user input:
try:
    num1 = input('Enter first number: ')
    num2 = input('Enter second number: ')
except (EOFError, KeyboardInterrupt):
    # Fallback default values for demonstration
    num1, num2 = '10.5', '20.25'
    print(f"(Using default values: {num1}, {num2})")

# Add two numbers
sum = float(num1) + float(num2)

# Display the sum
print('The sum of {0} and {1} is {2}'.format(num1, num2, sum))

```

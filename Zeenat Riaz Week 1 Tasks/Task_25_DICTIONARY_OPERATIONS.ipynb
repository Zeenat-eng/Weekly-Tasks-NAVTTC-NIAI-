###  TASK 25 DICTIONARY OPERATIONS


```python
"""
Task 25: Exercise on Dictionary Data Type in Python
Week: 1
Reference: https://www.programiz.com/python-programming/dictionary
"""

# 1. Dictionary Creation
student = {
    "roll_no": "AI-2026-001",
    "name": "Zeenat Riaz",
    "course": "Artificial Intelligence & Machine Learning",
    "semester": 1,
    "marks": {"Python": 95, "Math": 90, "AI": 92}
}
print("Original Student Dictionary:")
print(student)

# 2. Accessing Items
print(f"\nAccess by key ['name']: {student['name']}")
print(f"Access via .get('course'): {student.get('course')}")
print(f"Access missing key with default: {student.get('grade', 'Not Assigned')}")

# 3. Adding and Updating Items
student["city"] = "Islamabad"
student["semester"] = 2
print(f"\nAfter adding city & updating semester: {student}")

# 4. Removing Items
popped_city = student.pop("city")
print(f"Popped 'city': {popped_city}")

# 5. Iterating through Dictionary
print("\nIterating Keys and Values:")
for key, value in student.items():
    if key != "marks":
        print(f"  {key:<12}: {value}")

print("\nNested Dictionary (Marks):")
for subject, mark in student["marks"].items():
    print(f"  {subject:<10}: {mark}")

# 6. Dictionary Comprehension
numbers = [1, 2, 3, 4, 5]
cubes_dict = {x: x ** 3 for x in numbers}
print(f"\nCubes dictionary comprehension: {cubes_dict}")

```

###  TASK 24 SET OPERATIONS


```python
"""
Task 24: Exercise on Set Data Type in Python
Week: 1
Reference: https://www.programiz.com/python-programming/set
"""

# 1. Set Creation (Unordered, Unique elements)
number_set = {1, 2, 3, 4, 4, 3, 2, 1}
print(f"Set automatically removes duplicates: {number_set}")

# 2. Adding and Removing Elements
skills = {"Python", "NumPy", "Pandas"}
skills.add("Matplotlib")
skills.update(["Scikit-Learn", "TensorFlow"])
print(f"Set after adding items: {skills}")

skills.remove("NumPy")      # Raises KeyError if not present
skills.discard("Unknown")   # Does NOT raise error if missing
print(f"Set after remove/discard: {skills}")

# 3. Mathematical Set Operations
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print(f"\nSet A: {set_a}")
print(f"Set B: {set_b}")
print(f"Union (A | B)        : {set_a | set_b}")
print(f"Intersection (A & B) : {set_a & set_b}")
print(f"Difference (A - B)   : {set_a - set_b}")
print(f"Difference (B - A)   : {set_b - set_a}")
print(f"Symmetric Diff (A ^ B): {set_a ^ set_b}")

# 4. Set Membership & Subset Checking
subset_candidate = {1, 2}
print(f"\nIs {subset_candidate} subset of Set A? {subset_candidate.issubset(set_a)}")
print(f"Is Set A superset of {subset_candidate}? {set_a.issuperset(subset_candidate)}")

```

###  TASK 18 CONTINUE STATEMENT


```python
"""
Task 18: Exercise on Continue Statement in Python
Week: 1
Reference: https://www.geeksforgeeks.org/python-continue-statement/
"""

print("--- 1. Continue to Print Only Odd Numbers ---")
for num in range(1, 11):
    if num % 2 == 0:
        continue  # Skip even numbers
    print(f"Odd number: {num}")

print("\n--- 2. Continue in String Iteration (Skipping Vowels) ---")
sentence = "Artificial Intelligence"
vowels = "aeiouAEIOU"
consonants_only = []

for char in sentence:
    if char in vowels or char == " ":
        continue
    consonants_only.append(char)

print(f"Original: {sentence}")
print(f"Consonants: {''.join(consonants_only)}")

print("\n--- 3. Continue in a While Loop ---")
count = 0
while count < 6:
    count += 1
    if count == 3:
        print("Skipping iteration when count == 3")
        continue
    print(f"Count: {count}")

```

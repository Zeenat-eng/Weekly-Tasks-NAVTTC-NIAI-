###  TASK 22 TUPLE OPERATIONS


```python
"""
Task 22: Exercise on Tuple Data Type in Python
Week: 1
Reference: https://www.programiz.com/python-programming/tuple
"""

# 1. Tuple Creation
empty_tuple = ()
single_element_tuple = ("Python",)  # Trailing comma is essential
student_tuple = ("Alice", 24, "Computer Science", 3.85)
print(f"Student tuple: {student_tuple}")
print(f"Single element tuple type: {type(single_element_tuple)}")

# 2. Accessing Tuple Elements (Indexing & Slicing)
print(f"Name: {student_tuple[0]}")
print(f"GPA: {student_tuple[-1]}")
print(f"Slice [1:3]: {student_tuple[1:3]}")

# 3. Tuple Immutability Demonstration
print("\nDemonstrating Tuple Immutability:")
try:
    student_tuple[1] = 25  # Will raise TypeError
except TypeError as e:
    print(f"  Caught expected error: {e}")

# 4. Tuple Unpacking
name, age, department, gpa = student_tuple
print(f"\nUnpacked: Name={name}, Age={age}, Dept={department}, GPA={gpa}")

# 5. Tuple Methods (count & index)
letters = ('a', 'b', 'c', 'b', 'd', 'b')
print(f"\nCount of 'b': {letters.count('b')}")
print(f"First index of 'c': {letters.index('c')}")

# 6. Converting between List and Tuple
temp_list = list(student_tuple)
temp_list[1] = 25  # Can modify list
updated_tuple = tuple(temp_list)
print(f"Updated tuple via list conversion: {updated_tuple}")

```

###  TASK 23 STRING OPERATIONS


```python
"""
Task 23: Exercise on String Data Type in Python
Week: 1
Reference: https://www.programiz.com/python-programming/string
"""

# 1. String Creation and Basic Attributes
text = "Artificial Intelligence & Machine Learning"
print(f"Original text: '{text}'")
print(f"Length of string: {len(text)}")

# 2. String Indexing and Slicing
print(f"First character: '{text[0]}'")
print(f"Last character: '{text[-1]}'")
print(f"First word (slice 0:10): '{text[0:10]}'")
print(f"Reversed string: '{text[::-1]}'")

# 3. Essential String Methods
sample = "  hello deep learning world!  "
print(f"Stripped whitespace: '{sample.strip()}'")
print(f"Uppercase: '{sample.upper()}'")
print(f"Title Case: '{sample.title()}'")
print(f"Replace 'deep' with 'machine': '{sample.replace('deep', 'machine')}'")

# 4. Splitting and Joining
words = text.split(" ")
print(f"Splitting by space: {words}")
joined = " - ".join(words)
print(f"Joining with dash: '{joined}'")

# 5. String Validation Methods
digit_str = "2026"
alpha_str = "NAVTTC"
print(f"Is '{digit_str}' digit? {digit_str.isdigit()}")
print(f"Is '{alpha_str}' alpha? {alpha_str.isalpha()}")

# 6. String Formatting (f-string and .format())
course_id = "CS-401"
enrolled = 45
print(f"Course {course_id} has {enrolled} students enrolled. (f-string)")
print("Course {0} has {1} students enrolled. (.format)".format(course_id, enrolled))

```

###  TASK 15 FOR LOOPS


```python
"""
Task 15: Exercise on For Loops in Python
Week: 1
Reference: https://www.geeksforgeeks.org/python-for-loops/
"""

print("--- 1. Iterating Over a List ---")
fruits = ["apple", "banana", "cherry", "mango"]
for fruit in fruits:
    print(f"Fruit: {fruit}")

print("\n--- 2. Iterating Over a String ---")
word = "Python"
for char in word:
    print(char, end=" ")
print()

print("\n--- 3. Using range(start, stop, step) ---")
for i in range(1, 10, 2):
    print(f"Odd number: {i}")

print("\n--- 4. For Loop with else Block ---")
# The else block executes when the loop finishes without hitting a break
for n in range(3):
    print(f"Current count: {n}")
else:
    print("Loop completed successfully!")

print("\n--- 5. Nested For Loops (Multiplication Table Preview) ---")
for r in range(1, 4):
    for c in range(1, 4):
        print(f"{r}x{c}={r*c:2d}", end="  ")
    print()

print("\n--- 6. List Comprehension (Concise For Loop) ---")
squares = [x ** 2 for x in range(1, 6)]
print(f"Squares of 1 to 5: {squares}")

```

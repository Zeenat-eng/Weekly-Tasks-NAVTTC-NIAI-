###  TASK 19 LOOPING TECHNIQUES


```python
"""
Task 19: Exercise on Various Looping Techniques in Python
Week: 1
Reference: https://www.geeksforgeeks.org/looping-techniques-python/
"""

print("--- 1. Using enumerate() ---")
technologies = ["Python", "Machine Learning", "Deep Learning", "Azure AI"]
for index, tech in enumerate(technologies, start=1):
    print(f"{index}. {tech}")

print("\n--- 2. Using zip() to Iterate Multiple Sequences ---")
names = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]
grades = ["B", "A", "C"]
for name, score, grade in zip(names, scores, grades):
    print(f"Student: {name:<8} Score: {score}  Grade: {grade}")

print("\n--- 3. Using reversed() ---")
numbers = [10, 20, 30, 40, 50]
print("Reversed sequence: ", end="")
for val in reversed(numbers):
    print(val, end=" ")
print()

print("\n--- 4. Using sorted() without modifying original ---")
unsorted_list = [5, 2, 9, 1, 5, 6]
print("Sorted order: ", end="")
for val in sorted(unsorted_list):
    print(val, end=" ")
print(f"\nOriginal preserved: {unsorted_list}")

print("\n--- 5. Dictionary Iteration Techniques ---")
course_info = {"Name": "AI & ML", "Duration": "3 Months", "Status": "Active"}
print("Keys and Values via .items():")
for key, val in course_info.items():
    print(f"  {key:<10}: {val}")

```

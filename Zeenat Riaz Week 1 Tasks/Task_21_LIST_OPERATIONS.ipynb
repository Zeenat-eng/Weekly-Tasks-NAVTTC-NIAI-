###  TASK 21 LIST OPERATIONS


```python
"""
Task 21: Exercise on List Data Type in Python
Week: 1
Reference: https://www.programiz.com/python-programming/list
"""

# 1. List Creation
fruits = ["apple", "banana", "cherry", "date"]
mixed_list = [101, "Python", 3.14, True]
print(f"Original fruits list: {fruits}")

# 2. Accessing Elements (Indexing & Slicing)
print(f"First element [0]: {fruits[0]}")
print(f"Last element [-1]: {fruits[-1]}")
print(f"Slice [1:3]: {fruits[1:3]}")

# 3. Adding Elements
fruits.append("elderberry")         # Add to end
fruits.insert(1, "blueberry")       # Insert at specific index
more_fruits = ["fig", "grape"]
fruits.extend(more_fruits)          # Extend with another list
print(f"List after appends/inserts: {fruits}")

# 4. Modifying Elements
fruits[0] = "avocado"
print(f"After modifying first item: {fruits}")

# 5. Removing Elements
fruits.remove("banana")             # Remove by value
popped_item = fruits.pop()          # Remove last item
del fruits[1]                       # Delete at index 1
print(f"Popped item: {popped_item}")
print(f"List after removals: {fruits}")

# 6. List Methods
numbers = [45, 12, 89, 23, 7, 12]
print(f"Number of occurrences of 12: {numbers.count(12)}")
print(f"Index of 89: {numbers.index(89)}")
numbers.sort()
print(f"Sorted numbers: {numbers}")
numbers.reverse()
print(f"Reversed numbers: {numbers}")

# 7. List Comprehension
even_squares = [x ** 2 for x in range(1, 11) if x % 2 == 0]
print(f"Squares of even numbers 1-10: {even_squares}")

```

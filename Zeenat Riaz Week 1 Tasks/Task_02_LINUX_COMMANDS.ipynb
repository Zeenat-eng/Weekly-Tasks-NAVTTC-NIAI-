```python
"""
Task 2: Linux Commands Reference and Demonstration
Course: Artificial Intelligence, Machine Learning & Deep Learning
Week: 1

Practice the essential Linux/Debian shell commands:
pwd, cd, ls, cat, sudo, man, redirection (> >> <), mkdir, rm, rmdir, cp, mv,
file, reading, more, less, head, alias, shutdown, restart, touch, nano,
bash, sh, chmod, ps, kill, dpkg
"""

import os
import shutil

LINUX_COMMANDS_REFERENCE = {
    "pwd": "Print Working Directory - shows current directory path. Example: pwd",
    "cd": "Change Directory - navigate between folders. Example: cd /home/user or cd ..",
    "ls": "List directory contents. Options: ls -la (all files with details), ls -lh",
    "cat": "Concatenate & display file content. Example: cat file.txt",
    "sudo": "Execute command with superuser (administrator) privileges. Example: sudo apt update",
    "man": "Display system manual/documentation for any command. Example: man ls",
    "redirection (> >> <)": "Redirect stdout/stdin. '>' overwrites, '>>' appends. Example: echo 'hello' > a.txt",
    "mkdir": "Create new directory. Example: mkdir -p my_folder/sub_folder",
    "rm": "Remove files or directories. Example: rm file.txt or rm -rf my_folder",
    "rmdir": "Remove empty directory. Example: rmdir empty_dir",
    "cp": "Copy files or directories. Example: cp source.txt dest.txt or cp -r dir1 dir2",
    "mv": "Move or rename files/directories. Example: mv old.txt new.txt or mv file.txt /tmp/",
    "file": "Determine file type. Example: file image.png",
    "more": "View file content page-by-page (forward only). Example: more large.log",
    "less": "View file content with backward and forward navigation. Example: less large.log",
    "head": "Output first lines of files (default 10). Example: head -n 5 file.txt",
    "alias": "Create a shortcut for commands. Example: alias ll='ls -la'",
    "shutdown": "Shutdown system safely. Example: sudo shutdown now",
    "restart": "Reboot system. Example: sudo reboot or sudo shutdown -r now",
    "touch": "Create empty file or update timestamp. Example: touch script.py",
    "nano": "Simple command-line text editor. Example: nano script.py",
    "bash": "GNU Bourne-Again SHell executable. Example: bash script.sh",
    "sh": "Standard POSIX command shell interpreter. Example: sh script.sh",
    "chmod": "Change file permissions. Example: chmod +x script.sh (or chmod 755)",
    "ps": "Report process status snapshot. Options: ps aux, ps -ef",
    "kill": "Terminate process by PID. Example: kill -9 1234",
    "dpkg": "Debian package manager tool. Example: sudo dpkg -i package.deb"
}

def display_linux_guide():
    print("=" * 70)
    print("      DEBIAN / LINUX ESSENTIAL SHELL COMMANDS REFERENCE")
    print("=" * 70)
    for cmd, desc in LINUX_COMMANDS_REFERENCE.items():
        print(f"[*] {cmd:<22} : {desc}")
    print("=" * 70)

def python_simulation_demo():
    print("\nDemonstrating basic file & folder operations in Python (cross-platform equivalent):")
    demo_dir = "demo_linux_practice"
    os.makedirs(demo_dir, exist_ok=True)
    print(f"1. [mkdir] Created directory: {demo_dir}")

    demo_file = os.path.join(demo_dir, "sample.txt")
    with open(demo_file, "w") as f:
        f.write("Line 1: Introduction to Linux\nLine 2: AI and ML Fundamentals\nLine 3: End of file.\n")
    print(f"2. [touch/redirection] Created file: {demo_file}")

    print("3. [cat/head] Reading file contents:")
    with open(demo_file, "r") as f:
        print(f.read().strip())

    # Cleanup demo
    shutil.rmtree(demo_dir)
    print(f"4. [rm -rf] Cleaned up demo directory: {demo_dir}")

if __name__ == "__main__":
    display_linux_guide()
    python_simulation_demo()

```

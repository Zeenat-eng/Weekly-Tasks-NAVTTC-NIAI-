```python
"""
Task 26: Exercise on Exception Handling in Python
Week: 2
Reference: https://www.programiz.com/python-programming/exception-handling
"""

print("=== 1. Basic try-except block ===")
try:
    numerator = 10
    denominator = 0
    result = numerator / denominator
    print(f"Result: {result}")
except ZeroDivisionError as e:
    print(f"Error caught: Cannot divide by zero! ({e})")

print("\n=== 2. Handling Multiple Specific Exceptions ===")
test_inputs = ["abc", 0, 5]
for val in test_inputs:
    try:
        divisor = int(val)
        res = 100 / divisor
        print(f"100 / {divisor} = {res}")
    except ValueError as ve:
        print(f"ValueError: '{val}' is not a valid integer. ({ve})")
    except ZeroDivisionError as zde:
        print(f"ZeroDivisionError: Division by zero attempted. ({zde})")

print("\n=== 3. Using try-except-else-finally ===")
def divide_numbers(a, b):
    print(f"\nAttempting to divide {a} by {b}:")
    try:
        ans = a / b
    except ZeroDivisionError:
        print("  [EXCEPT] Error: Denominator cannot be zero.")
    else:
        # Executes only if no exception occurred in try block
        print(f"  [ELSE] Success: Calculation completed! Result = {ans}")
    finally:
        # Executes ALWAYS, regardless of exceptions
        print("  [FINALLY] Cleanup: Exiting divide_numbers function.")

divide_numbers(20, 4)
divide_numbers(10, 0)

print("\n=== 4. Raising an Exception Manually ===")
def validate_percentage(score):
    if score < 0 or score > 100:
        raise ValueError(f"Score {score} is invalid! Must be between 0 and 100.")
    return f"Valid score: {score}%"

try:
    print(validate_percentage(85))
    print(validate_percentage(120))
except ValueError as err:
    print(f"Custom validation caught: {err}")

```

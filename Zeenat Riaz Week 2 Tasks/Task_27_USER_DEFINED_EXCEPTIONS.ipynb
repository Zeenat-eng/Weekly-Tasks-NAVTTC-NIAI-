###  TASK 27 USER DEFINED EXCEPTIONS


```python
"""
Task 27: Exercise on User-Defined Exception Handling in Python
Week: 2
Reference: https://www.programiz.com/python-programming/user-defined-exception
"""

# 1. Defining Custom Exception Classes inheriting from built-in Exception
class InvalidAgeError(Exception):
    """Raised when the input age is negative or unrealistically high."""
    def __init__(self, age, message="Age must be between 18 and 100 for AI training eligibility"):
        self.age = age
        self.message = f"{message} (Provided age: {age})"
        super().__init__(self.message)

class InsufficientBalanceError(Exception):
    """Raised when an account withdrawal exceeds available balance."""
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        self.message = f"Cannot withdraw ${amount}. Current balance is only ${balance}."
        super().__init__(self.message)

# 2. Testing Custom Age Exception
def register_candidate(name, age):
    print(f"\nRegistering candidate: {name}, Age: {age}")
    if age < 18 or age > 100:
        raise InvalidAgeError(age)
    print(f"Candidate {name} successfully registered!")

# 3. Testing Custom Banking Exception
class BankAccount:
    def __init__(self, account_title, initial_balance):
        self.title = account_title
        self.balance = initial_balance

    def withdraw(self, amount):
        print(f"\nAccount '{self.title}' withdrawal request: ${amount}")
        if amount > self.balance:
            raise InsufficientBalanceError(self.balance, amount)
        self.balance -= amount
        print(f"Withdrawal successful. Remaining balance: ${self.balance}")

if __name__ == "__main__":
    # Test InvalidAgeError
    for test_age in [22, 15]:
        try:
            register_candidate("Ali Khan", test_age)
        except InvalidAgeError as ex:
            print(f"  [ERROR] Registration Failed -> {ex}")

    # Test InsufficientBalanceError
    account = BankAccount("AI Project Fund", 500)
    for withdraw_amount in [200, 450]:
        try:
            account.withdraw(withdraw_amount)
        except InsufficientBalanceError as ex:
            print(f"  [ERROR] Transaction Declined -> {ex}")

```

###  TASK 50 STANDARD DEVIATION


```python
"""
Task 50: Descriptive Statistics - Standard Deviation and Variance
Week: 2
Reference: https://www.w3schools.com/python/python_ml_standard_deviation.asp
"""

import math
import statistics
import numpy as np

# Sample dataset (e.g. ages of individuals in a research cohort)
ages = [32, 28, 37, 35, 30, 26, 34, 38, 29, 31]

print("=" * 60)
print("     MEASURES OF DISPERSION: VARIANCE & STANDARD DEVIATION")
print("=" * 60)
print(f"Dataset (Ages): {ages}")
print(f"Count (N)     : {len(ages)}")
print("-" * 60)

# --- Step-by-Step Manual Calculation ---
mean_val = sum(ages) / len(ages)
squared_deviations = [(x - mean_val) ** 2 for x in ages]
# Population variance divides by N; Sample variance divides by (N - 1)
pop_variance_manual = sum(squared_deviations) / len(ages)
sample_variance_manual = sum(squared_deviations) / (len(ages) - 1)
pop_std_manual = math.sqrt(pop_variance_manual)
sample_std_manual = math.sqrt(sample_variance_manual)

print(f"Mean of dataset       : {mean_val:.2f}")
print(f"\n--- VARIANCE (Average of squared differences from mean) ---")
print(f"Population Variance (N)  : {pop_variance_manual:.2f} (NumPy: {np.var(ages):.2f})")
print(f"Sample Variance (N - 1)  : {sample_variance_manual:.2f} (statistics.variance: {statistics.variance(ages):.2f})")

print(f"\n--- STANDARD DEVIATION (Square root of variance) ---")
print(f"Population Std Dev (N)   : {pop_std_manual:.2f} (NumPy: {np.std(ages):.2f})")
print(f"Sample Std Dev (N - 1)   : {sample_std_manual:.2f} (statistics.stdev: {statistics.stdev(ages):.2f})")

print("\nINTERPRETATION:")
print("A low standard deviation means data points are clustered closely around the mean.")
print("A high standard deviation indicates that the values are spread out over a wider range.")
print("=" * 60)

```

```python
"""
Task 49: Descriptive Statistics - Mean, Median, and Mode
Week: 2
Reference: https://www.w3schools.com/python/python_ml_mean_median_mode.asp
"""

import statistics
import numpy as np

# Sample dataset (e.g. speeds of 13 cars registered by radar)
speed_data = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

print("=" * 60)
print("     MEASURES OF CENTRAL TENDENCY: MEAN, MEDIAN, MODE")
print("=" * 60)
print(f"Dataset: {speed_data}")
print(f"Number of observations (N): {len(speed_data)}")
print("-" * 60)

# --- 1. MEAN (Average value) ---
# Formula: Sum of all values / Count
manual_mean = sum(speed_data) / len(speed_data)
stats_mean = statistics.mean(speed_data)
numpy_mean = np.mean(speed_data)

print(f"1. MEAN (Average):")
print(f"   - Manual formula  : {manual_mean:.2f}")
print(f"   - statistics.mean : {stats_mean:.2f}")
print(f"   - numpy.mean      : {numpy_mean:.2f}")

# --- 2. MEDIAN (Middle value when sorted) ---
# If N is odd: middle element. If N is even: average of two middle elements.
sorted_data = sorted(speed_data)
print(f"\n2. MEDIAN (Middle Value):")
print(f"   - Sorted dataset   : {sorted_data}")
stats_median = statistics.median(speed_data)
numpy_median = np.median(speed_data)
print(f"   - statistics.median: {stats_median}")
print(f"   - numpy.median     : {numpy_median}")

# --- 3. MODE (Most frequently occurring value) ---
stats_mode = statistics.mode(speed_data)
print(f"\n3. MODE (Most Frequent Value):")
print(f"   - statistics.mode  : {stats_mode} (appears {speed_data.count(stats_mode)} times)")

print("=" * 60)

```
